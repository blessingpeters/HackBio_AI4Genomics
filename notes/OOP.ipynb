# The init Function
This is the most important OOP concept.

````python
ecoli = Bacteria()
````

````python
class Bacteria:

    def __init__(self, species):
        self.species = species
````

# What is self?
This confuses everyone.

Don't overthink it.

Suppose:

```bash
ecoli = Bacteria("E.coli")
```
Python internally thinks:
```bash
ecoli.species = "E.coli"
```
The word:
```bash
self
```
simply means:
```bash
THIS OBJECT
```
Nothing more.



````python
class Bacteria:

    def __init__(self, species):
        self.species = species
````

means:

Store species inside THIS bacterium

Multiple Attributes
Let's add more information.

````python
class Bacteria:

    def __init__(self, species, growth_rate):
        self.species = species
        self.growth_rate = growth_rate
````

Create object:

````python
ecoli = Bacteria("E.coli", 0.8)
````

````python
print(ecoli.species)
print(ecoli.growth_rate)
````

# Methods
Attributes store information.

Methods perform actions.

Biological analogy:

Species = Property

Grow() = Action

````python
class Bacteria:

    def __init__(self, species):
        self.species = species

    def grow(self):
        print(self.species, "is growing")
````

````python
ecoli = Bacteria("E.coli")
````

````python
ecoli.grow()
````

# Adding More Biological Behavior

````python
class Bacteria:

    def __init__(self, species):
        self.species = species

    def divide(self):
        print(self.species, "is dividing")
````

````python
ecoli = Bacteria("E.coli")

ecoli.divide()
````

# Example: Cell Culture
Let's simulate cell growth.

````python
class CellCulture:

    def __init__(self, cell_count):
        self.cell_count = cell_count

    def grow(self):
        self.cell_count = self.cell_count * 2

        print("Cells:", self.cell_count)
````

````python
culture = CellCulture(100)
````

````python
culture.grow()
````

````python
#Grow again:

culture.grow()
````

# Inheritance
Now imagine:
```bash
Animal
├── Dog
├── Cat
└── Human
```
All are animals.

Instead of rewriting everything:
```bash
class Animal:
    ...

 ```
then:
```bash
class Dog(Animal):
    ...
```
Dog inherits Animal functionality.

````python
class Animal:

    def breathe(self):
        print("Breathing")
````

````python
class Dog(Animal):
    pass
````

````python
dog = Dog()
dog.breathe()


# Even though Dog never defined it. It inherited it.

````

# Why Inheritance Matters in PyTorch
Every neural network inherits from:

```bash
nn.Module
```
Example:
```bash
class SimpleCNN(nn.Module):
    ...
```
This means:

SimpleCNN
inherits
nn.Module

PyTorch gives us:
model training
parameter tracking
model saving
evaluation mode
for free.



# super()
When inheriting, we must initialize the parent.

Example:



````python
class SimpleCNN(nn.Module):

    def __init__(self):

        super().__init__()
````

Think:

Dear Parent Class,

Please initialize yourself first.


then:

````python
super().__init__()
````

## Putting It All Together
This is essentially how every PyTorch model starts.

````python
import torch.nn as nn

class SimpleCNN(nn.Module):

    def __init__(self):

        super().__init__()

        self.fc = nn.Linear(10, 2)

    def forward(self, x):

        return self.fc(x)
````

At this stage you do not need to understand:

Linear layer
Forward pass
Neural networks
Just observe the OOP structure.
```bash
Class
│
├── Inheritance
│
├── Constructor (__init__)
│
├── Attributes (self.fc)
│
└── Methods (forward)
```
Everything else in PyTorch is built on these ideas.
